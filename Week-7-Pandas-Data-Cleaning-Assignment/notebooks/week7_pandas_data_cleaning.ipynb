**Objective:** Learn Python basics and perform basic data exploration and cleaning using Pandas.

This notebook uses a Superstore sales dataset and completes all required assignment steps: loading data, exploring it, handling missing values, filtering/selecting data, removing duplicates, creating a derived column, and saving the cleaned CSV output.

## 1. Import Libraries and Set File Paths

This cell imports Pandas and creates the required project folders so the notebook can save outputs without any path error.

```python
from pathlib import Path
import pandas as pd

# Base path is the main assignment folder.
# This works when the notebook is opened from the Week-7 assignment folder in VS Code.
BASE_DIR = Path.cwd()
if BASE_DIR.name == "notebooks":
    BASE_DIR = BASE_DIR.parent

RAW_DATA_PATH = BASE_DIR / "data" / "raw" / "superstore_sales.csv"
CLEANED_DATA_PATH = BASE_DIR / "data" / "processed" / "cleaned_superstore_sales.csv"
OUTPUT_DIR = BASE_DIR / "outputs"

# Create folders if they do not already exist
(BASE_DIR / "data" / "processed").mkdir(parents=True, exist_ok=True)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Base folder:", BASE_DIR)
print("Raw data path:", RAW_DATA_PATH)
print("Cleaned data path:", CLEANED_DATA_PATH)
```

## 2. Load CSV Dataset into Pandas DataFrame

The raw CSV file is loaded into a Pandas DataFrame using `pd.read_csv()`.

```python
df = pd.read_csv(RAW_DATA_PATH)
print("Dataset loaded successfully")
df.head()
```

## 4. Check and Handle Missing Values

Missing values are checked column-wise. Then practical cleaning rules are applied:

- Text columns are filled with `Unknown`.
- Numeric columns are filled with median values.
- Date columns are converted into datetime format.

```python
# Check missing values before cleaning
missing_before = df.isnull().sum()
missing_before[missing_before > 0]
```

```python
# Create a copy so the original raw data remains unchanged
cleaned_df = df.copy()

# Remove extra spaces from column names
cleaned_df.columns = cleaned_df.columns.str.strip()

# Convert date columns to datetime format
for date_col in ["Order Date", "Ship Date"]:
    cleaned_df[date_col] = pd.to_datetime(cleaned_df[date_col], errors="coerce", format="mixed")

# Convert important numeric columns to proper numeric data type
numeric_columns = ["Sales", "Quantity", "Discount", "Profit"]
for col in numeric_columns:
    cleaned_df[col] = pd.to_numeric(cleaned_df[col], errors="coerce")

# Fill missing text values with 'Unknown'
text_columns = cleaned_df.select_dtypes(include="object").columns
for col in text_columns:
    cleaned_df[col] = cleaned_df[col].fillna("Unknown")

# Fill missing numeric values with median
number_columns = cleaned_df.select_dtypes(include="number").columns
for col in number_columns:
    cleaned_df[col] = cleaned_df[col].fillna(cleaned_df[col].median())

# Fill missing dates if any date parsing failed
cleaned_df["Order Date"] = cleaned_df["Order Date"].fillna(pd.Timestamp("1900-01-01"))
cleaned_df["Ship Date"] = cleaned_df["Ship Date"].fillna(cleaned_df["Order Date"])

print("Missing values after cleaning:", cleaned_df.isnull().sum().sum())
```

## 5. Basic Operations: Filter Rows and Select Columns

This step performs simple Pandas operations by filtering profitable Technology orders and selecting important columns.

```python
# Filter rows: Technology category orders where profit is positive
technology_profitable_orders = cleaned_df[(cleaned_df["Category"] == "Technology") & (cleaned_df["Profit"] > 0)]
technology_profitable_orders.head()
```

```python
# Select useful columns for a smaller preview table
selected_columns = cleaned_df[["Order ID", "Customer Name", "Category", "Sub-Category", "Sales", "Quantity", "Profit"]]
selected_columns.head()
```

## 6. Remove Duplicates

Duplicate rows are checked and removed using `drop_duplicates()`.

```python
duplicates_before = cleaned_df.duplicated().sum()
print("Duplicate rows before cleaning:", duplicates_before)

cleaned_df = cleaned_df.drop_duplicates()

duplicates_after = cleaned_df.duplicated().sum()
print("Duplicate rows after cleaning:", duplicates_after)
```

## 7. Create Derived Column: `total_amount = price * quantity`

The provided dataset does not have a direct `price` column. It has `Sales` and `Quantity`.

So, `price` is calculated as:

`price = Sales / Quantity`

Then the required derived column is created as:

`total_amount = price * Quantity`

```python
# Create price column from Sales and Quantity
cleaned_df["price"] = (cleaned_df["Sales"] / cleaned_df["Quantity"]).round(2)

# Create total_amount column using the assignment formula
cleaned_df["total_amount"] = (cleaned_df["price"] * cleaned_df["Quantity"]).round(2)

cleaned_df[["Sales", "Quantity", "price", "total_amount"]].head()
```

## 8. Save Cleaned Dataset as New CSV

The final cleaned dataset is saved inside the `data/processed/` folder.

```python
cleaned_df.to_csv(CLEANED_DATA_PATH, index=False)
print("Cleaned dataset saved successfully at:", CLEANED_DATA_PATH)
```

## 9. Save Extra Output Files

These files make the assignment folder easier to review because they show the filtered data and summary results separately.

```python
# Save filtered Technology orders
technology_profitable_orders.to_csv(OUTPUT_DIR / "technology_profitable_orders.csv", index=False)

# Save selected columns preview
selected_columns.head(100).to_csv(OUTPUT_DIR / "selected_columns_preview.csv", index=False)

# Category-level summary
category_summary = cleaned_df.groupby("Category", as_index=False).agg(
    total_sales=("Sales", "sum"),
    total_profit=("Profit", "sum"),
    total_quantity=("Quantity", "sum"),
    unique_orders=("Order ID", "nunique")
).sort_values("total_sales", ascending=False)

category_summary.to_csv(OUTPUT_DIR / "category_summary.csv", index=False)
category_summary
```

## 10. Final Summary

This final cell prints a short summary of the cleaned dataset and output files.

```python
print("Final cleaned dataset shape:", cleaned_df.shape)
print("Total sales:", round(cleaned_df["Sales"].sum(), 2))
print("Total profit:", round(cleaned_df["Profit"].sum(), 2))
print("Total quantity sold:", int(cleaned_df["Quantity"].sum()))
print("Unique orders:", cleaned_df["Order ID"].nunique())
print("Output files saved inside:", OUTPUT_DIR)
```
